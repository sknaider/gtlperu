## 2. Generate Synthetic Dataset

```python
from backend.modules.datasets.generator import DatasetGenerator

# Create generator
generator = DatasetGenerator()

# Generate classification dataset
dataset = generator.generate_text_classification_dataset(
    task_description="Sentiment analysis of product reviews",
    categories=["positive", "negative", "neutral"],
    num_samples=50  # Start small
)

print(f"Generated {len(dataset)} samples")
print("\nExample:", dataset[0])
```

## 3. RAG System

```python
from backend.modules.rag import RAGSystem

# Create RAG system
rag = RAGSystem(collection_name="quickstart")

# Index some documents
documents = [
    "Machine learning is a subset of artificial intelligence.",
    "Deep learning uses neural networks with many layers.",
    "PyTorch is a popular deep learning framework.",
    "The RTX 5090 is NVIDIA's latest GPU for AI workloads.",
]

rag.index_documents(documents)
print(f"Indexed {len(documents)} documents")
```

```python
# Query the RAG system
result = rag.query("What is deep learning?")

print("Answer:", result["answer"])
print("\nSources:")
for i, source in enumerate(result["sources"], 1):
    print(f"{i}. {source['text']} (score: {source['score']:.3f})")
```

## 4. Image Classification

```python
from backend.modules.vision.classifier import create_classifier
from PIL import Image
import requests
from io import BytesIO

# Create classifier
classifier = create_classifier("efficientnet_b0", pretrained=True)

# Load example image
url = "https://upload.wikimedia.org/wikipedia/commons/3/3a/Cat03.jpg"
response = requests.get(url)
image = Image.open(BytesIO(response.content))

# Predict
predictions = classifier.predict(image, top_k=5)

print("Top 5 Predictions:")
for class_id, prob in predictions:
    print(f"  Class {class_id}: {prob*100:.2f}%")
```
